```python
%load_ext autoreload
%autoreload 2

import pandas as pd
import pdfplumber
import re
import os
from utils.utils_oab import extract_text, find_middle, extract_by_columns, join_texts, find_question_pages, load_rules, clean_text, split_questions, extract_questions, extract_answer_key
idx = 0
print(f'Prova: {exams_list[idx]}')
exam_path = f'all_exams_oab/{exams_list[idx]}.pdf'
answer_key_path = f'all_exams_oab/{exams_list[idx]}[gabarito].pdf'
#edicao = exams_list[idx].split("Exame da Ordem Unificado")[1].split(" - ")[0].strip()
size_pdf = 0
with pdfplumber.open(exam_path) as pdf:
    size_pdf = len(pdf.pages)
literals, patterns = load_rules()
```

```python
dict_mid = {i: 0.5 for i in range(size_pdf)}
# Column split (fraction of the page width) of the pages whose right column does not start at the middle
custom_mid = {
    'CESPE - 2010 - OAB - Exame de Ordem Unificado - I - Primeira Fase': {
        1: 0.496, 2: 0.496, 3: 0.496, 4: 0.511, 5: 0.511, 6: 0.511, 7: 0.487, 8: 0.487,
        9: 0.487, 10: 0.511, 11: 0.511, 12: 0.514, 13: 0.514, 14: 0.481, 15: 0.466,
    },
}
dict_mid.update(custom_mid.get(exams_list[idx], {}))
meio = find_middle(exam_path,dict_mid)
```

```python
aa_left, aa_right = extract_by_columns(exam_path, dict_mid)
first_page, last_page = find_question_pages(aa_left, aa_right, literals, patterns)
print(f'Question pages: {first_page} to {last_page} (of {len(aa_left)})')
aa_left = aa_left[first_page:last_page + 1]
aa_right = aa_right[first_page:last_page + 1]
```

```python
questions = join_texts(aa_left, aa_right)
```

```python
questions = clean_text(questions, literals, patterns)
```

```python
questions_split = split_questions(questions)
```

```python
question_literals, question_patterns = load_rules(stage="question")
list_questions = extract_questions(questions_split, question_literals, question_patterns)
```

```python
extracted_questions = extract_questions(questions_split, question_literals, question_patterns)
```

```python
answers_path = f'{exam_path.split(".")[0]}[gabarito].pdf'
```

```python
gabarito_text = []
with pdfplumber.open(answers_path) as pdf:
    for pagina in pdf.pages:
        gabarito_text.append(pagina.extract_text())
```

```python
answer_key = extract_answer_key(answer_key_path)
if len(answer_key) != len(extracted_questions):
    print(f'Answer key has {len(answer_key)} answers, but {len(extracted_questions)} questions were extracted.')
```

```python
# answer is None for annulled questions
for question_data in extracted_questions:
    question_data['answer'] = answer_key.get(question_data['question_number'])
```

```python
csv_path = 'csvs/oab/oab_questions_all.csv'
df_exam = pd.DataFrame([{
    'question_number': question_data['question_number'],
    'question': question_data['question'],
    'options': list(question_data['options'].values()),
    'exam': exams_list[idx],
    'answer': question_data['answer'],
} for question_data in extracted_questions])

# Questions typed manually (unreadable in the PDF) are appended in question order
manual_path = 'csvs/oab/oab_questions_manual.csv'
df_manual = pd.read_csv(manual_path, encoding='utf-8-sig')
df_manual = df_manual[df_manual['exam'] == exams_list[idx]]
if len(df_manual):
    df_exam = pd.concat([df_exam, df_manual], ignore_index=True).sort_values('question_number', ignore_index=True)
    print(f'{len(df_manual)} manual questions added: {df_manual["question_number"].tolist()}')

if os.path.exists(csv_path):
    df_saved = pd.read_csv(csv_path, encoding='utf-8-sig')
    df_saved = df_saved[df_saved['exam'] != exams_list[idx]]
    df_exam = pd.concat([df_saved, df_exam], ignore_index=True)
df_exam.to_csv(csv_path, index=False, encoding='utf-8-sig')
print(f'{csv_path}: {len(df_exam)} questions from {df_exam["exam"].nunique()} exams')
```
